```python
from __future__ import annotations
from pathlib import Path
import re, json
from collections import Counter

SECRET_RE = re.compile(r"\$\{\{\s*secrets\.", re.IGNORECASE)
PUBLISH_RE = re.compile(
    r"(docker/login-action|npm\s+publish|pypi|twine\s+upload|"
    r"configure-aws-credentials|google-github-actions/auth|azure/login|"
    r"codecov|gh-release|release|publish|deploy|build-push-action)",
    re.IGNORECASE,
)
FORK_HOSTILE_RE = re.compile(r"(pull_request_target)", re.IGNORECASE)

def scan_workflow_text(text: str) -> dict:
    uses_secrets = bool(SECRET_RE.search(text))
    publish_or_deploy = bool(PUBLISH_RE.search(text))
    pr_target = bool(FORK_HOSTILE_RE.search(text))
    return {
        "uses_secrets": uses_secrets,
        "publish_or_deploy": publish_or_deploy,
        "pull_request_target": pr_target,
        "likely_secret_dependent": (uses_secrets or publish_or_deploy),
    }

def scan_repo(repo_path: Path) -> dict:
    wf_dir = repo_path / ".github" / "workflows"
    results = {"repo_path": str(repo_path), "workflows": []}

    if not wf_dir.exists():
        results["status"] = "no_workflows"
        return results

    any_secret_dep = False
    any_pr_target = False

    for wf in sorted(wf_dir.glob("*.y*ml")):
        text = wf.read_text(errors="ignore")
        hits = scan_workflow_text(text)
        results["workflows"].append({"file": str(wf.relative_to(repo_path)), **hits})
        any_secret_dep |= hits["likely_secret_dependent"]
        any_pr_target |= hits["pull_request_target"]

    if any_secret_dep:
        results["status"] = "skip_secret_dependent"
    elif any_pr_target:
        results["status"] = "review_pull_request_target"
    else:
        results["status"] = "ok_to_fork_and_run"

    return results

def find_git_repos(root: Path) -> list[Path]:
    # Find every ".git" marker (dir or file), and treat its parent as a repo root.
    repo_roots = []
    seen = set()

    for git_marker in root.rglob(".git"):
        repo = git_marker.parent
        if repo in seen:
            continue
        # sanity: ensure it looks like a repo root
        if (repo / ".git").exists():
            seen.add(repo)
            repo_roots.append(repo)

    return sorted(repo_roots)

root = Path(r"./cloned_repos")  # change if needed
repos = find_git_repos(root)

print(f"Found {len(repos)} git repos under {root}")

report = []
for repo in repos:
    report.append(scan_repo(repo))

Path("workflow_scan_report.json").write_text(json.dumps(report, indent=2))

c = Counter(r["status"] for r in report)
print("Counts:", c)

ok = [r for r in report if r["status"] == "ok_to_fork_and_run"]
Path("ok_to_fork_and_run.json").write_text(json.dumps(ok, indent=2))
print(f"OK-to-fork count: {len(ok)}")

with open("ok_to_fork_and_run.json") as f:
    ok_repos = json.load(f)

```

```python
import subprocess, time, random

ORG = "gha-workflow-study"

def fork_repo(src_full: str, org: str = ORG, max_retries: int = 8) -> tuple[bool, str]:
    """
    Fork a repo into an org, handling GitHub secondary rate limits ("submitted too quickly").
    Returns (ok, message).
    """
    base_delay = 1.5  # steady pacing between requests even on success

    backoff = 10.0    # starting backoff when throttled
    # for attempt in range(max_retries):
    while True:
        # small jitter BEFORE each attempt to avoid synchronized bursts
        time.sleep(base_delay + random.uniform(0, 0.8))

        p = subprocess.run(
            ["gh", "repo", "fork", src_full, "--org", org, "--clone=false"],
            capture_output=True,
            text=True,
        )

        if p.returncode == 0:
            return True, p.stdout.strip()

        err = ((p.stderr or "") + "\n" + (p.stdout or "")).strip()
        err_l = err.lower()

        throttled = (
            "submitted too quickly" in err_l
            or "secondary rate limit" in err_l
            or "abuse" in err_l
            or "rate limit" in err_l
        )

        if not throttled:
            # Not a throttling issue; don’t spin forever.
            return False, err

        # Backoff w/ jitter (cap at 10 minutes)
        sleep_s = min(backoff, 600.0) + random.uniform(0, 3.0)
        print(f"[THROTTLED] {src_full} -> sleeping {sleep_s:.1f}s (backoff {backoff:.1f}s)")
        # print(f"[THROTTLED] {src_full} -> sleeping {sleep_s:.1f}s (attempt {attempt+1}/{max_retries})")
        time.sleep(sleep_s)
        backoff *= 2.0

    return False, f"gave up after {max_retries} retries: {src_full}"

# for r in ok_repos:
for r in ok_repos_filtered:
    if r["status"] != "ok_to_fork_and_run":
        continue

    owner, repo = r["repo_path"].split("\\")[-2:]
    src = f"{owner}/{repo}"
    print(src, end=" ")

    ok, msg = fork_repo(src)
    print("OK" if ok else "FAIL")
    if ok == "OK":
        successful_forks.append(src)
    if not ok:
        print(msg.splitlines()[0][:300])
```

```python
from __future__ import annotations

import argparse
import csv
import json
import os
import subprocess
import time
from pathlib import Path
from typing import Optional
import sys

max_int = sys.maxsize
while True:
    try:
        csv.field_size_limit(max_int)
        break
    except OverflowError:
        max_int = int(max_int / 10)

def sh(cmd: list[str], cwd: Optional[Path] = None, timeout: int = 600) -> str:
    r = subprocess.run(cmd, cwd=str(cwd) if cwd else None, capture_output=True, text=True, timeout=timeout)
    if r.returncode != 0:
        raise RuntimeError(
            f"CMD failed: {' '.join(cmd)}\nCWD: {cwd}\nRC: {r.returncode}\nSTDERR:\n{r.stderr}\nSTDOUT:\n{r.stdout}"
        )
    return r.stdout.strip()


def gh_json(args: list[str], timeout: int = 600):
    out = sh(["gh", *args], timeout=timeout)
    out_stripped = out.strip()
    if not out_stripped:
        raise RuntimeError(f"gh returned empty output for: gh {' '.join(args)}")
    try:
        return json.loads(out_stripped)
    except json.JSONDecodeError as exc:
        raise RuntimeError(
            "Failed to parse JSON from gh output.\n"
            f"CMD: gh {' '.join(args)}\n"
            f"RAW OUTPUT:\n{out_stripped}\n"
        ) from exc

def sh2(cmd: list[str], cwd: Optional[Path] = None, timeout: int = 600) -> tuple[int, str, str]:
    r = subprocess.run(
        cmd,
        cwd=str(cwd) if cwd else None,
        capture_output=True,
        text=True,
        timeout=timeout
    )
    return r.returncode, (r.stdout or ""), (r.stderr or "")


def parse_json_loose(raw: str) -> object:
    """
    Attempts to parse JSON even if raw contains leading noise (warnings, banners).
    Strategy:
      1) try normal json.loads on stripped text
      2) if that fails, find first '{' or '[' and try from there
    """
    s = raw.strip()
    if not s:
        raise RuntimeError("Empty output (no JSON to parse).")

    try:
        return json.loads(s)
    except json.JSONDecodeError:
        pass

    # try to salvage if gh printed a warning/banner before JSON
    i_obj = s.find("{")
    i_arr = s.find("[")
    starts = [i for i in (i_obj, i_arr) if i != -1]
    if not starts:
        raise RuntimeError(f"Output is not JSON. First 500 chars:\n{s[:500]}")
    i = min(starts)

    try:
        return json.loads(s[i:])
    except json.JSONDecodeError as exc:
        raise RuntimeError(
            "Failed to parse JSON even after trimming leading noise.\n"
            f"First 500 chars:\n{s[:500]}\n"
            f"--- Trimmed-from index {i} ---\n{s[i:i+500]}"
        ) from exc


def gh_json_api(endpoint: str, timeout: int = 600):
    # --silent reduces progress/noise; -H ensures JSON response format
    cmd = ["gh", "api", "--silent", "-H", "Accept: application/vnd.github+json", endpoint]

    rc, out, err = sh2(cmd, timeout=timeout)

    # gh sometimes prints important info to stderr even when rc==0
    if rc != 0:
        raise RuntimeError(
            f"CMD failed: {' '.join(cmd)}\nRC: {rc}\nSTDERR:\n{err}\nSTDOUT:\n{out}"
        )

    if not out.strip():
        raise RuntimeError(
            "gh api returned empty stdout.\n"
            f"CMD: {' '.join(cmd)}\n"
            f"STDERR:\n{err.strip()}\n"
            "Common causes: not authenticated, SSO required, or gh is printing only to stderr."
        )

    return parse_json_loose(out)

def get_default_branch_local(repo_dir: Path) -> str:
    # Ensure we have origin refs
    sh(["git", "fetch", "origin", "--prune"], cwd=repo_dir)

    # This prints something like: "origin/main"
    ref = sh(["git", "symbolic-ref", "refs/remotes/origin/HEAD"], cwd=repo_dir)
    # ref = "refs/remotes/origin/main" or "origin/main" depending on git; normalize:
    ref = ref.strip()

    # Common forms:
    #  - refs/remotes/origin/main
    #  - origin/main
    if ref.startswith("refs/remotes/"):
        parts = ref.split("/")
        return parts[-1]  # last component is branch name
    if ref.startswith("origin/"):
        return ref.split("/", 1)[1]
    return ref


def read_workflow_file(repo_dir: Path, workflow_path: str) -> str:
    p = repo_dir / workflow_path
    return p.read_text(encoding="utf-8", errors="ignore")


def detect_trigger_strategy(workflow_yaml: str) -> str:
    # Bare-minimum heuristics (good enough to start):
    txt = workflow_yaml.lower()
    has_dispatch = "workflow_dispatch" in txt
    has_pr = "pull_request" in txt
    has_push = "\non:\n" in txt and "push" in txt  # rough, but ok for v1
    # Prefer dispatch if present
    if has_dispatch:
        return "dispatch"
    if has_pr:
        return "pr"
    if has_push:
        return "push"
    return "unsupported"


def ensure_branch(repo_dir: Path, branch: str, base_branch: str):
    # Create/reset branch to base_branch state locally
    sh(["git", "fetch", "origin", base_branch], cwd=repo_dir)
    sh(["git", "checkout", base_branch], cwd=repo_dir)
    sh(["git", "reset", "--hard", f"origin/{base_branch}"], cwd=repo_dir)
    sh(["git", "checkout", "-B", branch], cwd=repo_dir)


def push_branch(repo_dir: Path, branch: str):
    sh(["git", "push", "-u", "origin", branch, "--force"], cwd=repo_dir)


def commit_empty(repo_dir: Path, msg: str):
    sh(["git", "commit", "--allow-empty", "-m", msg], cwd=repo_dir)


def overwrite_workflow(repo_dir: Path, workflow_path: str, new_content: str):
    p = repo_dir / workflow_path
    p.parent.mkdir(parents=True, exist_ok=True)
    p.write_text(new_content, encoding="utf-8")


def commit_all(repo_dir: Path, msg: str):
    sh(["git", "add", "-A"], cwd=repo_dir)
    sh(["git", "commit", "-m", msg], cwd=repo_dir)


def trigger_dispatch(repo_full: str, workflow_path: str, ref: str):
    sh(["gh", "workflow", "run", workflow_path, "--repo", repo_full, "--ref", ref])


def create_pr(repo_full: str, head: str, base: str, title: str):
    # If PR exists already, gh may error. For v1, ignore if it already exists.
    try:
        sh(["gh", "pr", "create", "--repo", repo_full, "--head", head, "--base", base, "--title", title, "--body", ""])
    except RuntimeError as e:
        # If PR already exists or cannot be created, we still might get a run from pushing updates.
        print(f"[warn] PR create failed (may already exist). Continuing.\n{e}")


def latest_run(repo_full: str, branch: str, workflow_path: str):
    runs = gh_json([
        "run", "list",
        "--repo", repo_full,
        "--branch", branch,
        "--workflow", workflow_path,
        "--limit", "5",
        "--json", "databaseId,status,conclusion,event,createdAt,workflowName,headBranch"
    ]) or []
    return runs[0] if runs else None


def wait_for_run_complete(repo_full: str, branch: str, workflow_path: str, timeout_s: int = 1800, poll_s: int = 8):
    start = time.time()
    while time.time() - start < timeout_s:
        run = latest_run(repo_full, branch, workflow_path)
        if run:
            if run["status"] == "completed":
                return run
        time.sleep(poll_s)
    raise TimeoutError(f"Run did not complete in time: {repo_full} {branch} {workflow_path}")


def download_log(repo_full: str, run_id: int, out_path: Path):
    out_path.parent.mkdir(parents=True, exist_ok=True)
    log = sh(["gh", "run", "view", str(run_id), "--repo", repo_full, "--log"], timeout=600)
    out_path.write_text(log, encoding="utf-8")


def download_jobs(repo_full: str, run_id: int, out_path: Path):
    out_path.parent.mkdir(parents=True, exist_ok=True)
    jobs = gh_json(["run", "view", str(run_id), "--repo", repo_full, "--json", "jobs"], timeout=600)
    out_path.write_text(json.dumps(jobs, indent=2), encoding="utf-8")


def classify_outcome(run: dict, jobs_json: dict) -> str:
    # Run exists but no jobs => effectively "skipped/no_jobs"
    jobs = (jobs_json or {}).get("jobs", []) if isinstance(jobs_json, dict) else []
    if not jobs:
        # could be skipped before jobs were created
        return "no_jobs"
    concl = run.get("conclusion")
    if concl == "success":
        return "success"
    if concl == "skipped":
        return "skipped"
    if concl:
        return concl
    return "unknown"


def main(argv=None):
    ap = argparse.ArgumentParser()
    ap.add_argument("--csv", required=True)
    ap.add_argument("--row", type=int, default=0)
    ap.add_argument("--repos-root", required=True)
    ap.add_argument("--org", required=True)
    ap.add_argument("--workflow-col", default="file_path")
    ap.add_argument("--yaml-col", default="generated_workflow")
    ap.add_argument("--timeout", type=int, default=1800)
    args = ap.parse_args(argv)

    # with open(args.csv, newline="", encoding="utf-8") as f:
    #     rows = list(csv.DictReader(f))
    # row = rows[args.row]

    with open(args.csv, newline="", encoding="utf-8") as f:
        reader = csv.DictReader(f)
        for i, row in enumerate(reader):
            if i == args.row:
                break
        else:
            raise ValueError(f"Row {args.row} not found")

    repo_rel = row["repository"]                  # "0xax/linux-insides"
    repo_name = repo_rel.split("/", 1)[1]         # "linux-insides"
    repo_dir = Path(args.repos_root) / repo_name   # "./forked_repos/0xax/linux-insides"
    repo_dir = "./" + str(repo_dir).replace("\\", "/")  # normalize for Windows
    print(repo_dir)
    repo_full = f"{args.org}/{repo_name}"         # "gha-workflow-study/linux-insides"

    workflow_path = row[args.workflow_col]        # ".github/workflows/check-code-snippets.yaml"
    gen_yaml = row[args.yaml_col]                 # YAML text

    base_branch = get_default_branch_local(repo_dir)
    print("default branch:", base_branch)
    print("workflow path:", workflow_path)

    # Inspect current workflow to pick trigger strategy
    wf_text = read_workflow_file(repo_dir, workflow_path)
    strategy = detect_trigger_strategy(wf_text)
    print("trigger strategy:", strategy)
    if strategy == "unsupported":
        print("unsupported trigger; stopping.")
        return

    out_root = Path("stage3_out") / repo_full.replace("/", "__") / Path(workflow_path).name
    out_root.mkdir(parents=True, exist_ok=True)

    # ---------- BASELINE ----------
    ensure_branch(repo_dir, "baseline", base_branch)

    # For push/PR triggers we need a new commit to generate the event.
    if strategy in ("push", "pr"):
        commit_empty(repo_dir, "baseline trigger")
        push_branch(repo_dir, "baseline")

    if strategy == "dispatch":
        push_branch(repo_dir, "baseline")  # ensure ref exists remotely
        trigger_dispatch(repo_full, workflow_path, "baseline")
    elif strategy == "pr":
        # create PR baseline -> base_branch
        create_pr(repo_full, head="baseline", base=base_branch, title="baseline")
    elif strategy == "push":
        # already pushed above; that should trigger if workflow listens to push for that branch
        pass

    run_b = wait_for_run_complete(repo_full, "baseline", workflow_path, timeout_s=args.timeout)
    download_log(repo_full, run_b["databaseId"], out_root / "baseline" / "log.txt")
    download_jobs(repo_full, run_b["databaseId"], out_root / "baseline" / "jobs.json")
    jobs_b = json.loads((out_root / "baseline" / "jobs.json").read_text(encoding="utf-8"))
    outcome_b = classify_outcome(run_b, jobs_b)
    print("baseline outcome:", outcome_b, "run_id:", run_b["databaseId"])

    # ---------- GENERATED ----------
    ensure_branch(repo_dir, "generated", base_branch)

    overwrite_workflow(repo_dir, workflow_path, gen_yaml)
    commit_all(repo_dir, "swap in generated workflow")
    push_branch(repo_dir, "generated")

    if strategy == "dispatch":
        trigger_dispatch(repo_full, workflow_path, "generated")
    elif strategy == "pr":
        create_pr(repo_full, head="generated", base=base_branch, title="generated")
    elif strategy == "push":
        # push event already happened; add another empty commit to trigger again
        commit_empty(repo_dir, "generated trigger")
        push_branch(repo_dir, "generated")

    run_g = wait_for_run_complete(repo_full, "generated", workflow_path, timeout_s=args.timeout)
    download_log(repo_full, run_g["databaseId"], out_root / "generated" / "log.txt")
    download_jobs(repo_full, run_g["databaseId"], out_root / "generated" / "jobs.json")
    jobs_g = json.loads((out_root / "generated" / "jobs.json").read_text(encoding="utf-8"))
    outcome_g = classify_outcome(run_g, jobs_g)
    print("generated outcome:", outcome_g, "run_id:", run_g["databaseId"])

    summary = {
        "repo": repo_full,
        "workflow_path": workflow_path,
        "strategy": strategy,
        "baseline": {"run_id": run_b["databaseId"], "conclusion": run_b.get("conclusion"), "outcome": outcome_b},
        "generated": {"run_id": run_g["databaseId"], "conclusion": run_g.get("conclusion"), "outcome": outcome_g},
    }
    (out_root / "summary.json").write_text(json.dumps(summary, indent=2), encoding="utf-8")
    print("wrote:", out_root / "summary.json")


main([
  "--csv", "../ML Ops Final Project Backup/workflows_dataset_rq2_complete.csv",
  "--repos-root", "./forked_repos/gha_workflow_study/",
  "--org", "gha-workflow-study",
  "--row", "0",
  "--workflow-col", "file_path",
  "--yaml-col", "generated_workflow",
])

```
